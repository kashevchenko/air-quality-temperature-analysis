```python
# ============================================================
# ИССЛЕДОВАНИЕ ВЗАИМОСВЯЗИ ТЕМПЕРАТУРЫ И ЗАГРЯЗНЕНИЯ ВОЗДУХА
# CORRELATION + LINEAR REGRESSION + POLYNOMIAL REGRESSION
# ============================================================

!pip -q install pandas numpy matplotlib seaborn scipy scikit-learn statsmodels

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import pearsonr, spearmanr
from statsmodels.stats.multitest import multipletests
import statsmodels.api as sm

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

import warnings
warnings.filterwarnings("ignore")

# ============================================================
# 1. ЗАГРУЗКА ДАННЫХ
# ============================================================

FILE_PATH = "/content/air_quality_dataset.csv"

df = pd.read_csv(FILE_PATH)

print("=" * 65)
print("ИССЛЕДОВАНИЕ ТЕМПЕРАТУРЫ И ЗАГРЯЗНЕНИЯ ВОЗДУХА")
print("=" * 65)

print("\nРазмер датасета:", df.shape)

print("\nСтолбцы:")
print(df.columns.tolist())

display(df.head())

# ============================================================
# 2. ОПРЕДЕЛЕНИЕ ПЕРЕМЕННЫХ
# ============================================================

temperature = "Temperature"

pollutants = [
    "PM2.5",
    "PM10",
    "NOx",
    "NO2",
    "SO2",
    "VOCs",
    "CO",
    "CO2",
    "CH4"
]

required_columns = [temperature] + pollutants

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"В файле отсутствуют столбцы: {missing_columns}"
    )

# Преобразование в числовой формат
for col in required_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

# Удаление пропущенных значений
df = df.dropna(
    subset=required_columns
).copy()

# ============================================================
# 3. ПРОВЕРКА КАЧЕСТВА ДАННЫХ
# ============================================================

print("\nПРОВЕРКА КАЧЕСТВА ДАННЫХ")

print("Количество наблюдений:", len(df))

print(
    "Количество дубликатов:",
    df.duplicated().sum()
)

negative_values = (df[pollutants] < 0).sum()

print("\nОтрицательные значения концентраций:")
display(negative_values.to_frame("Количество"))

# Отрицательные концентрации физически невозможны.
# Для основного анализа исключаем такие наблюдения.

df_clean = df[
    (df[pollutants] >= 0).all(axis=1)
].copy()

print(
    "\nНаблюдений после очистки:",
    len(df_clean)
)

# ============================================================
# 4. ОПИСАТЕЛЬНАЯ СТАТИСТИКА
# ============================================================

print("\nОПИСАТЕЛЬНАЯ СТАТИСТИКА")

display(
    df_clean[required_columns].describe().T.round(3)
)

# ============================================================
# 5. КОРРЕЛЯЦИЯ ПИРСОНА И СПИРМЕНА
# ============================================================

correlation_results = []

for pollutant in pollutants:

    x = df_clean[temperature]
    y = df_clean[pollutant]

    pearson_r, pearson_p = pearsonr(x, y)

    spearman_r, spearman_p = spearmanr(x, y)

    correlation_results.append({
        "Pollutant": pollutant,
        "Pearson_r": pearson_r,
        "Pearson_p": pearson_p,
        "Spearman_r": spearman_r,
        "Spearman_p": spearman_p
    })

corr_df = pd.DataFrame(correlation_results)

# Поправка на множественные проверки:
# Benjamini-Hochberg для 9 тестов Пирсона.

corr_df["Pearson_p_FDR"] = multipletests(
    corr_df["Pearson_p"],
    method="fdr_bh"
)[1]

corr_df["Significant_FDR"] = (
    corr_df["Pearson_p_FDR"] < 0.05
)

corr_df = corr_df.sort_values(
    "Pearson_r",
    ascending=False
)

print("\nКОРРЕЛЯЦИОННЫЙ АНАЛИЗ")

display(corr_df.round(5))

# ============================================================
# 6. ВИЗУАЛИЗАЦИЯ КОРРЕЛЯЦИЙ
# ============================================================

plt.figure(figsize=(12, 7))

sns.barplot(
    data=corr_df,
    x="Pearson_r",
    y="Pollutant",
    color="steelblue"
)

plt.axvline(
    x=0,
    color="black",
    linestyle="--"
)

plt.title(
    "Корреляция температуры с загрязняющими веществами",
    fontsize=15
)

plt.xlabel("Коэффициент корреляции Пирсона")
plt.ylabel("Загрязнитель")

plt.xlim(-1, 1)

plt.show()

# ============================================================
# 7. ПОЛНАЯ КОРРЕЛЯЦИОННАЯ МАТРИЦА
# ============================================================

correlation_matrix = df_clean[
    required_columns
].corr(method="pearson")

plt.figure(figsize=(12, 9))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5
)

plt.title(
    "Корреляционная матрица загрязнителей и температуры"
)

plt.show()

# ============================================================
# 8. ЛИНЕЙНАЯ РЕГРЕССИЯ
# ============================================================

linear_results = []

for pollutant in pollutants:

    X = df_clean[[temperature]]
    y = df_clean[pollutant]

    X_sm = sm.add_constant(X)

    model = sm.OLS(
        y,
        X_sm
    ).fit(cov_type="HC3")

    linear_results.append({
        "Pollutant": pollutant,
        "Intercept": model.params["const"],
        "Temperature_Coefficient": model.params[temperature],
        "P_value": model.pvalues[temperature],
        "R_squared": model.rsquared,
        "CI_lower": model.conf_int().loc[temperature, 0],
        "CI_upper": model.conf_int().loc[temperature, 1]
    })

linear_df = pd.DataFrame(linear_results)

print("\nРЕЗУЛЬТАТЫ ЛИНЕЙНОЙ РЕГРЕССИИ")

display(linear_df.round(5))

# ============================================================
# 9. ГРАФИКИ ЛИНЕЙНОЙ РЕГРЕССИИ
# ============================================================

fig, axes = plt.subplots(
    3,
    3,
    figsize=(18, 15)
)

axes = axes.flatten()

for i, pollutant in enumerate(pollutants):

    sns.regplot(
        data=df_clean,
        x=temperature,
        y=pollutant,
        ax=axes[i],
        scatter_kws={
            "alpha": 0.4
        },
        line_kws={
            "color": "red"
        }
    )

    axes[i].set_title(
        f"Temperature vs {pollutant}"
    )

plt.tight_layout()
plt.show()

# ============================================================
# 10. СРАВНЕНИЕ LINEAR И POLYNOMIAL REGRESSION
# ============================================================

X = df_clean[[temperature]]

# Один и тот же train/test split
# для всех загрязнителей.

train_idx, test_idx = train_test_split(
    np.arange(len(df_clean)),
    test_size=0.25,
    random_state=42
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

comparison_results = []

for pollutant in pollutants:

    y = df_clean[pollutant]

    y_train = y.iloc[train_idx]
    y_test = y.iloc[test_idx]

    # Линейная регрессия
    linear_model = LinearRegression()

    linear_model.fit(
        X_train,
        y_train
    )

    linear_pred = linear_model.predict(X_test)

    # Полиномиальная регрессия второй степени
    poly_model = make_pipeline(
        PolynomialFeatures(
            degree=2,
            include_bias=False
        ),
        LinearRegression()
    )

    poly_model.fit(
        X_train,
        y_train
    )

    poly_pred = poly_model.predict(X_test)

    comparison_results.append({
        "Pollutant": pollutant,

        "Linear_R2": r2_score(
            y_test,
            linear_pred
        ),

        "Polynomial_R2": r2_score(
            y_test,
            poly_pred
        ),

        "Linear_RMSE": np.sqrt(
            mean_squared_error(
                y_test,
                linear_pred
            )
        ),

        "Polynomial_RMSE": np.sqrt(
            mean_squared_error(
                y_test,
                poly_pred
            )
        ),

        "Linear_MAE": mean_absolute_error(
            y_test,
            linear_pred
        ),

        "Polynomial_MAE": mean_absolute_error(
            y_test,
            poly_pred
        )
    })

comparison_df = pd.DataFrame(comparison_results)

comparison_df["Best_Model"] = np.where(
    comparison_df["Polynomial_RMSE"]
    < comparison_df["Linear_RMSE"],
    "Polynomial",
    "Linear"
)

print("\nСРАВНЕНИЕ МОДЕЛЕЙ")

display(comparison_df.round(4))

# ============================================================
# 11. ВИЗУАЛИЗАЦИЯ СРАВНЕНИЯ МОДЕЛЕЙ
# ============================================================

comparison_plot = comparison_df.melt(
    id_vars="Pollutant",
    value_vars=[
        "Linear_R2",
        "Polynomial_R2"
    ],
    var_name="Model",
    value_name="R2"
)

plt.figure(figsize=(13, 7))

sns.barplot(
    data=comparison_plot,
    x="Pollutant",
    y="R2",
    hue="Model"
)

plt.title(
    "Сравнение R² линейной и полиномиальной моделей"
)

plt.ylabel("R² на тестовой выборке")
plt.xlabel("Загрязнитель")

plt.axhline(
    0,
    color="black",
    linewidth=0.8
)

plt.show()

# ============================================================
# 12. ВИЗУАЛИЗАЦИЯ НЕЛИНЕЙНОЙ ЗАВИСИМОСТИ
# ============================================================

# Выберем CO2 как пример.

TARGET = "CO2"

X = df_clean[[temperature]]
y = df_clean[TARGET]

linear_model = LinearRegression()

linear_model.fit(X, y)

poly_model = make_pipeline(
    PolynomialFeatures(
        degree=2,
        include_bias=False
    ),
    LinearRegression()
)

poly_model.fit(X, y)

temperature_range = np.linspace(
    X[temperature].min(),
    X[temperature].max(),
    200
).reshape(-1, 1)

temperature_grid = pd.DataFrame(
    temperature_range,
    columns=[temperature]
)

linear_curve = linear_model.predict(
    temperature_grid
)

poly_curve = poly_model.predict(
    temperature_grid
)

plt.figure(figsize=(12, 7))

plt.scatter(
    X[temperature],
    y,
    alpha=0.4,
    label="Наблюдения"
)

plt.plot(
    temperature_range,
    linear_curve,
    label="Linear Regression",
    linewidth=2
)

plt.plot(
    temperature_range,
    poly_curve,
    label="Polynomial Regression",
    linewidth=2
)

plt.title(
    "Сравнение линейной и нелинейной зависимости: Temperature → CO2"
)

plt.xlabel("Temperature")
plt.ylabel("CO2")

plt.legend()
plt.show()

# ============================================================
# 13. АВТОМАТИЧЕСКИЕ ВЫВОДЫ
# ============================================================

print("\n" + "=" * 70)
print("ИТОГОВЫЕ ВЫВОДЫ ИССЛЕДОВАНИЯ")
print("=" * 70)

for _, row in corr_df.iterrows():

    pollutant = row["Pollutant"]

    r = row["Pearson_r"]

    p = row["Pearson_p_FDR"]

    if abs(r) < 0.2:
        strength = "очень слабая"

    elif abs(r) < 0.4:
        strength = "слабая"

    elif abs(r) < 0.6:
        strength = "умеренная"

    elif abs(r) < 0.8:
        strength = "сильная"

    else:
        strength = "очень сильная"

    direction = (
        "положительная"
        if r > 0
        else "отрицательная"
    )

    print(f"\n{pollutant}")

    print(
        f"Корреляция: {r:.3f}"
    )

    print(
        f"Сила связи: {strength}"
    )

    print(
        f"Направление: {direction}"
    )

    print(
        f"FDR-adjusted p-value: {p:.5f}"
    )

    if p < 0.05:

        print(
            "Статистически значимая линейная связь обнаружена."
        )

    else:

        print(
            "Статистически значимая линейная связь не обнаружена."
        )

# ============================================================
# 14. СОХРАНЕНИЕ РЕЗУЛЬТАТОВ
# ============================================================

corr_df.to_csv(
    "/content/temperature_correlations.csv",
    index=False,
    encoding="utf-8-sig"
)

linear_df.to_csv(
    "/content/temperature_linear_regression.csv",
    index=False,
    encoding="utf-8-sig"
)

comparison_df.to_csv(
    "/content/temperature_model_comparison.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 70)
print("АНАЛИЗ ЗАВЕРШЁН")
print("=" * 70)

print("\nСозданы файлы:")

print("temperature_correlations.csv")
print("temperature_linear_regression.csv")
print("temperature_model_comparison.csv")
```
